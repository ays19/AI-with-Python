# NumPy

```python
# why numpy

import time
import numpy as np

# creating large data
size = 1000000

# python list
list1 = list(range(size))
list2 = list(range(size))

# Numpy Array
arr1 = np.array(list1)
arr2 = np.array(list2)

# list addition time
start = time.time()

result_list =[]
for i in range(size):
    result_list.append(list1[i] + list2[i])

end = time.time()

print("List Time:", end - start)

# numpy addition time
start = time.time()

result_array = arr1 + arr2

end = time.time()

print("NumPy Time:", end - start)
```
