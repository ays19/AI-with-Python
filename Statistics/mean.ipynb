# Calculating Mean

# Method 1 - Manual Calculation

```python
# Population data
population_data = [150,160,170,180,145]

# Sample data
sample_data = [150,170,165]

# Calculate population mean
population_mean = sum(population_data)/ len(population_data)
print(f"Population Mean: {population_mean}")

# Calculate sample mean
sample_mean = sum(sample_data)/ len(sample_data)
print(f"Sample Mean: {sample_mean}")
```

# Method 2 - Using Numpy

```python
import numpy as np
# define the dataset
data = [80,90,20,70,30]

#calculate the mean using numpy
arithmetic_mean = np.mean(data)

# Output the result
print(f"Arithmentic Mean using numpy: {arithmetic_mean}")
```

# Method 3 - Using Staistics

```python
import statistics as stats

# sample dataset
data = [10,50,55,78,89]

#Calculate the mean using statistics module
mean_stats = stats.mean(data)

print(f"Mean using statistics module: {mean_stats}")
```

# Finding mean - Data set

```python
import seaborn as sns
import numpy as np

# Load the iris dataset
iris = sns.load_dataset("iris")

#Display the first few rows of the dataset
print(iris.head())

# Calculate the mean of the ;sepal_length' column
mean_sepal_length = np.mean(iris['sepal_length'])

# Output the result
print(f"The mean of the 'sepal_length' column is: {mean_sepal_length}")
```
