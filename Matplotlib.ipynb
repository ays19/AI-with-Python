# Matplotlib  (Visualization with Python)

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.plot(x,y)
plt.show()
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
c = 'r'

plt.plot(x,y,c)
plt.show()
```

```python
x = [5,7,19,5,8,9]
y = [10,6,8,2,19,24]

plt.plot(x,y)
plt.show()
```

```python
x = [5,2,9,4,7]

y = [10,5,8,4,2]

plt.plot(x,y)
plt.plot(y, marker = 'o')

plt.show()
```

# Bar Graph

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.bar(x,y)
plt.show()
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
c = ['b', 'r', 'g', 'y']

plt.bar(x,y, color = c)
plt.show()
```

```python
import numpy as np
```

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3,8,1,10])

plt.barh(x,y)
plt.show()
```

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3,8,1,10])

plt.bar(x,y, color='black')
plt.show()
```

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3,8,1,10])

plt.bar(x,y, color='#9CAF50')
plt.show()
```

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3,8,1,10])

plt.bar(x,y, width = 0.2)
plt.show()
```
