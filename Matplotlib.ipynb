# Matplotlib  (Visualization with Python)

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.plot(x,y)
plt.show()
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
c = 'r'

plt.plot(x,y,c)
plt.show()
```

```python
x = [5,7,19,5,8,9]
y = [10,6,8,2,19,24]

plt.plot(x,y)
plt.show()
```
