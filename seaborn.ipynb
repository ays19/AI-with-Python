```python
import speech_recognition as sr
import pyttsx3  #pyttsx3 here tts= text to speech and x3 = python3
import wikipedia

engine = pyttsx3.init()

def speak(text):
    engine.say(text)
    engine.runAndWait()

recognizer = sr.Recognizer()

with sr.Microphone() as source:
    print("Listening...")
    audio = recognizer.listen(source)

try:
    command = recognizer.recognize_google(audio)
    print("You said:", command)

    result = wikipedia.summary(command, sentences=2)

    print(result)
    speak(result)

except:
    speak("Sorry, Icould not understand")
```
